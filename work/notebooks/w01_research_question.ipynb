## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*
### Research question

Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring?

### Decision

The decision is which content pages should receive attention first when review capacity is limited.

### Action

A reviewer could use the ranked results to prioritize pages for actions such as refreshing content, expanding it, protecting a strong page, pruning a weak page, or monitoring it further.

### Cost of a wrong recommendation

A false positive could waste limited review time and effort on a page that does not need attention.

A false negative could cause an important page with a real opportunity or problem to be missed.

The goal is therefore to support prioritization, not to guarantee that a particular action will improve performance.
